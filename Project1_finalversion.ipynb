```python
def build_sequence_features(genome, gene_info, window=1000):
    rows = []

    for row1 in gene_info.itertuples(index=False):

        
        chrom_len=len(genome[row1.chr])

        TSS=math.floor((row1.TSS_start+row1.TSS_end)/2) ##we take the center of the TSS as reference for our proximal region
        TSS_lower = TSS-window
        TSS_upper = TSS+window

        TSS_lower=max(0,TSS_lower)
        TSS_upper=min(TSS_upper,chrom_len)

        sequence = genome[row1.chr][TSS_lower:TSS_upper].seq
        sequence = sequence.upper()
        reverse = sequence[::-1]


        #handle empty string and None
        if sequence:
            GC_content = float((sequence.count("G")+sequence.count("C"))/len(sequence))

            CpG_density = float(sequence.count("CG")/len(sequence))

            rows.append([row1.gene_name, GC_content, CpG_density])
        else:
            rows.append([row1.gene_name, 0.0, 0.0])
    

    df_features= pd.DataFrame(rows,columns=["gene_name", "GC_content", "CpG_density"])

    return df_features
```

```python
def build_distal_features(data, gene_info, bins= 3, proximal_window=40000, distal_window=200000):

    rows = []  # collect one feature row per gene outside the loop
    columns = ["gene_name"] + [f"bin_{i+1}" for i in range(bins)]


    for row1 in gene_info.itertuples(index=False):

        features = []
        features.append(row1.gene_name)

        bin_size= int((distal_window-proximal_window)/bins)

        TSS=math.floor((row1.TSS_start+row1.TSS_end)/2) ##we take the center of the TSS as reference for our proximal region
        proximal_lower = TSS-proximal_window
        proximal_upper = TSS+proximal_window

        chrom_len = data.chroms(str(row1.chr))

        for i in range(bins):


            lower_start = proximal_lower - (i+1)*bin_size
            lower_end = proximal_lower - i*bin_size

            lower_start = max(0, lower_start)
            lower_end = min(chrom_len, lower_end)

            if lower_end<=lower_start:
                lower_mean = 0.0
            else:
                vals_lower = np.array(data.values(row1.chr, lower_start, lower_end, numpy=True))

                if vals_lower.size == 0:
                    lower_mean = 0.0
                else:
                    lower_mean = float(np.nanmean(np.nan_to_num(vals_lower, nan=0.0)))

            upper_start = proximal_upper + i*bin_size
            upper_end = proximal_upper + (i+1)*bin_size

            upper_start = max(0, upper_start)
            upper_end = min(chrom_len, upper_end)

            if upper_end<=upper_start:
                upper_mean = 0.0
            else:
                vals_upper = np.array(data.values(row1.chr, upper_start, upper_end, numpy=True))

                if vals_upper.size == 0:
                    upper_mean = 0.0
                else:
                    upper_mean = float(np.nanmean(np.nan_to_num(vals_upper, nan=0.0)))

            bin_average = 0.5 * (upper_mean + lower_mean)

            features.append(bin_average)

        rows.append(features) ##add the gene and its features to the list of genes

    df_features= pd.DataFrame(rows,columns=columns)

    return df_features
    
```

```python
##Input
# data: pandas dataframe with cols ["chrom", "chromStart", "chromEnd", "signalValue"]
# gene_info ["gene_name", "chr", "gene_start", "gene_end", "TSS_start", "TSS_end", "strand"]
# window size, default is 40'000 kb

##Output
# Pandas data frame ["gene_name", "signalValue"] with the maximal peak for each gene within a +- region of 40k bp.

def build_proximal_features(data, gene_info, window=40000):

    rows = []

    peaks_per_chr = {c: df.reset_index(drop=True) for c, df in data.groupby("chrom", sort=False)}

    for row1 in gene_info.itertuples(index=False):
        TSS=math.floor((row1.TSS_start+row1.TSS_end)/2) ##we take the center of the TSS as reference for our proximal region
        proximal_lower = TSS-window
        proximal_upper = TSS+window

        peaks = peaks_per_chr.get(str(row1.chr))

        starts = peaks["chromStart"].to_numpy() ##faster compute when handling numpy vs pandas
        ends = peaks["chromEnd"].to_numpy()
        signals = peaks["signalValue"].to_numpy()

        candidates = (starts < proximal_upper) & (ends > proximal_lower)
        if candidates.any():
            rows.append((row1.gene_name,float(signals[candidates].max())))  
        else:
            rows.append((row1.gene_name,0.0))

        df_features= pd.DataFrame(rows,columns=["gene_name", "signalValue"])



    return df_features
```

```python
##Input: 
    #dfs is a list with all data frame features that are to be included
##Output:
    #Dataframe with all features used for prediction as columns and rows correspond to each gene

def build_feature_matrix(dfs):

    tables = []
    for i, d in enumerate(dfs):
        # make columns unique by adding an index-based suffix
        rename_map = {
            col: f"{col}_{i}"
            for col in d.columns
            if col != "gene_name"
        }

        renamed = (
            d.rename(columns=rename_map)
             .set_index("gene_name")
        )
        tables.append(renamed)

    # outer-join all data frames on gene_name
    X = reduce(lambda left, right: left.join(right, how="outer"), tables)

    return X

```
